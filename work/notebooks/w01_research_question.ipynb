## 2. The Question: Decision, Action, and Cost of a Wrong Call

### Decision

Which content pages should be reviewed first for refresh,
expansion, protection, pruning, or monitoring?

### Action

A content team could use the ranked output to decide which pages
to review first, rather than manually reviewing all 30,000 pages.

For each prioritized page, the output should include a suggested
action and clear reason codes so that a human can understand why
the page was prioritized.

### Cost of a Wrong Call

A false recommendation could waste content-review time or lead to
unnecessary changes to a page that is already performing well.

A missed recommendation could also mean that a page needing
attention is not reviewed soon enough.

Therefore, the system should support human review rather than
automatically make content changes.

### Why This Is Not Just "Train a Model"

The goal is not simply to maximize a model metric.

The goal is to produce a useful and interpretable review queue.
A transparent baseline should be compared with any ML approach,
and the final recommendations should include reasons that a
content team can understand.

The analysis should also distinguish association from causation:
a signal associated with declining performance does not prove
that changing that signal will improve the page.

not to automatically decide what should happen to a page.
